**Challenge 1**

In this challenge, we will be working with pokemon data. The data can be found here:

- https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv

```python
#libraries
import pandas as pd
import scipy.stats as st
import numpy as np
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv")
df
```

- We posit that Pokemons of type Dragon have, on average, more HP stats than Grass. Choose the propper test and, with 5% significance, comment your findings.

```python
dragon_pokemon = df[df["Type 1"]=="Dragon"]["HP"]
grass_pokemon = df[df["Type 2"]=="Grass"]["HP"]
```

```python
#Two Sample T-test

#Set the hypothesis:
#H0: mu Dragon_Stats <= Grass_Stats
#H1: mu Grass_Stats > Grass_Stats

#significance level = 0.05
```

```python
st.ttest_ind(dragon_pokemon, grass_pokemon, equal_var=False)
```

```python
#Because p_value is lower than significance level, we reject the null hypothesis, 
# this means that Grass pokemons, on average, have more HP stats than grass
```

- We posit that Legendary Pokemons have different stats (HP, Attack, Defense, Sp.Atk, Sp.Def, Speed) when comparing with Non-Legendary. Choose the propper test and, with 5% significance, comment your findings.


```python
#Two Sample T-test

#Set the hypothesis:
#H0: mu Legendary_Stats = Non_Legndary_Stats
#H1: mu Legendary_Stats != Non_Legndary_Stats

#significance level = 0.05
```

```python
# "HP" Stat
legendary_pokemon = df[df["Legendary"]==True]["HP"]
non_legendary_pokemon = df[df["Legendary"]==False]["HP"]

st.ttest_ind(legendary_pokemon, non_legendary_pokemon, equal_var=False)
```

```python
# "Attack" Stat
legendary_pokemon = df[df["Legendary"]==True]["Attack"]
non_legendary_pokemon = df[df["Legendary"]==False]["Attack"]

st.ttest_ind(legendary_pokemon, non_legendary_pokemon, equal_var=False)
```

```python
# "Defense" Stat
legendary_pokemon = df[df["Legendary"]==True]["Defense"]
non_legendary_pokemon = df[df["Legendary"]==False]["Defense"]

st.ttest_ind(legendary_pokemon, non_legendary_pokemon, equal_var=False)
```

```python
# "Sp. Atk" Stat
legendary_pokemon = df[df["Legendary"]==True]["Sp. Atk"]
non_legendary_pokemon = df[df["Legendary"]==False]["Sp. Atk"]

st.ttest_ind(legendary_pokemon, non_legendary_pokemon, equal_var=False)
```

```python
# "Sp. Def" Stat
legendary_pokemon = df[df["Legendary"]==True]["Sp. Def"]
non_legendary_pokemon = df[df["Legendary"]==False]["Sp. Def"]

st.ttest_ind(legendary_pokemon, non_legendary_pokemon, equal_var=False)
```

```python
# "Speed" Stat
legendary_pokemon = df[df["Legendary"]==True]["Speed"]
non_legendary_pokemon = df[df["Legendary"]==False]["Speed"]

st.ttest_ind(legendary_pokemon, non_legendary_pokemon, equal_var=False)
```

```python
# Because p_value is lower than significance level in all stats, we reject the null hypothesis, 
# this means that Legendary pokemons, on average, have a different Defense stats in all categories if compared with Non Legendary ones
```

In this challenge, we will be working with california-housing data. The data can be found here:
- https://raw.githubusercontent.com/data-bootcamp-v4/data/main/california_housing.csv

```python
df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/california_housing.csv")
df.head()
```

**We posit that houses close to either a school or a hospital are more expensive.**

- School coordinates (-118, 34)
- Hospital coordinates (-122, 37)

We consider a house (neighborhood) to be close to a school or hospital if the distance is lower than 0.50.

Hint:
- Write a function to calculate euclidean distance from each house (neighborhood) to the school and to the hospital.
- Divide your dataset into houses close and far from either a hospital or school.
- Choose the propper test and, with 5% significance, comment your findings.
 

```python
#Two Sample T-test

# H0: mean(median_house_value | longitude == -118.34) = mean(median_house_value | longitude != -118.34)
# H1: mean(median_house_value | longitude == -118.34) != mean(median_house_value | longitude != -118.34)

#significance level = 0.05
```

```python
# 1 -- Write a function to calculate euclidean distance from each house (neighborhood) to the school and to the hospital.

import numpy as np

def euclidean_distance(lon1, lat1, lon2, lat2):
    return np.sqrt((lon1 - lon2)**2 + (lat1 - lat2)**2)
```

```python
# 2 -- Divide your dataset into houses close and far from either a hospital or school.
lon_school = -118
lat_school = 34

lon_hospital = -122
lat_hospital = 37

df["dist_school"] = euclidean_distance(df["longitude"], df["latitude"],
                                       lon_school, lat_school)

df["dist_hospital"] = euclidean_distance(df["longitude"], df["latitude"],
                                         lon_hospital, lat_hospital)
```

```python
# 3-- Choose the propper test and, with 5% significance, comment your findings.
near = df[df["dist_school"] <= 1]["median_house_value"]
far  = df[df["dist_school"] > 1]["median_house_value"]

st.ttest_ind(near, far, equal_var=False)
```

```python
# Because p_value is lower than significance level, we reject the null hypothesis, 
# this means that houses close to either a school or a hospital are more expensive
```
