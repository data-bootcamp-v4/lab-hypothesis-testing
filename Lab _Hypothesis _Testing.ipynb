```python
# Import required libraries
import pandas as pd
import numpy as np
import scipy.stats as st
```

```python
# Load Pokémon dataset
pokemon_df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv")

housing_df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/california_housing.csv")

# Display first rows
pokemon_df.head()
```

# 
Challenge 1 – Pokémon Dataset

We want to test whether Dragon-type Pokémon have higher average HP than Grass-type Pokémon.
# Test Choice:
Two independent groups:
- Numeric variable (HP)

- Unknown population variance
➡️ Two-sample independent t-test

(one‑tailed test)
# Step 1: State the Hypotheses
**Null Hypothesis (H₀)**
The average HP of Dragon Pokémon is equal to the average HP of Grass Pokémon.

H₀​:μDragon​=μGrass​

**Alternative Hypothesis (H₁)**
Dragon Pokémon have a higher average HP than Grass Pokémon.

H₁​:μDragon​>μGrass​ 

```python
# Select HP values for Dragon and Grass Pokémon
dragon_hp = pokemon_df[pokemon_df["Type 1"] == "Dragon"]["HP"]
grass_hp = pokemon_df[pokemon_df["Type 1"] == "Grass"]["HP"]

# Display sample sizes
print("Dragon Pokémon:", len(dragon_hp))
print("Grass Pokémon:", len(grass_hp))
```

```python
# Check Mean HP

print("Average HP (Dragon):", dragon_hp.mean())
print("Average HP (Grass):", grass_hp.mean())
```

```python
# Two-sample t-test (one-tailed)
t_stat, p_val_two_tailed = st.ttest_ind(dragon_hp, grass_hp, equal_var=False)

# Convert to one-tailed p-value
p_value = p_val_two_tailed / 2

print("T-statistic:", t_stat)
print("One-tailed p-value:", p_value)
```

```python
alpha = 0.05

if (p_value < alpha) and (t_stat > 0):
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

# Conclusion
At the 5% significance level, we reject the null hypothesis.
There is strong statistical evidence that Dragon-type Pokémon have higher average HP than Grass-type Pokémon.

# Exercise 2
**Question**

*Do Legendary Pokémon have different battle statistics than Non-Legendary Pokémon?*

The statistics are:
- HP
- Attack
- Defense
- Sp. Atk
- Sp. Def
- Speed

# Step 1: Hypotheses

*For each statistic:*

Null Hypothesis:

The average statistic is the same for Legendary and Non-Legendary Pokémon.

H0:μLegendary=μNon−Legendary
	​

Alternative Hypothesis

The average statistic is different.

H1:μLegendary=μNon−Legendary
	
Because there are two independent groups, we again use an:

Independent Two-Sample T-Test (Welch's)

```python
# Step 2: Separate the six groups

legendary = pokemon_df[pokemon_df["Legendary"] == True]
non_legendary = pokemon_df[pokemon_df["Legendary"] == False]
```

```python
# step 3: Perform Tests

# List of statistics
stats = ["HP", "Attack", "Defense", "Sp. Atk", "Sp. Def", "Speed"]

alpha = 0.05

for stat in stats:

    t_stat, p_value = st.ttest_ind(
        legendary[stat],
        non_legendary[stat],
        equal_var=False
    )

    print(f"\n{stat}")
    print("-"*30)
    print("Legendary Mean:", legendary[stat].mean())
    print("Non-Legendary Mean:", non_legendary[stat].mean())
    print("T-statistic:", round(t_stat,3))
    print("P-value:", round(p_value,5))

    if p_value < alpha:
        print("Reject H₀")
    else:
        print("Fail to Reject H₀")
```

# (α = 0.05)
You will find p-values extremely small for all stats.

# Conclusion
At the 5% significance level, we reject the null hypothesis for all six stats.
Legendary Pokémon have significantly different (and generally higher) stats compared to non‑Legendary Pokémon.

### Interpretation

For every statistic:

- If p-value < 0.05:
    - There is a statistically significant difference between Legendary and Non-Legendary Pokémon.

- If p-value > 0.05:
    - There is no significant difference.

Typically, Legendary Pokémon have significantly higher averages across most battle statistics.

# Problem

We want to determine whether houses located near a school or hospital are more expensive.

# Coordinates:

- School: (-118, 34)

- Hospital: (-122, 37)

A house is considered close if the Euclidean distance is less than 0.50.

```python
# Step1: Distance Function

def distance(x1, y1, x2, y2):
    return np.sqrt((x1 - x2)**2 + (y1 - y2)**2)
```

```python
# STEP2: Compute distances

# Calculate Distance to School
housing_df["dist_school"] = distance(housing_df["longitude"], housing_df["latitude"], -118, 34)

# Calculate Distance to Hospital
housing_df["dist_hospital"] = distance(housing_df["longitude"], housing_df["latitude"], -122, 37)
```

```python
# Step 3 — Define “close”
# Close if distance < 0.50 to either school or hospital.

housing_df["close"] = ((housing_df["dist_school"] < 0.5) | (housing_df["dist_hospital"] < 0.5))
```

```python
# step 4: Check Counts
housing_df["close"].value_counts()
```

```python
# Step 5: Split groups

close_prices = housing_df[housing_df["close"] == True]["median_house_value"]
far_prices = housing_df[housing_df["close"] == False]["median_house_value"]
```

```python
# Step 6: Compare Average Prices
print("Average Close:", close_prices.mean())
print("Average Far:", far_prices.mean())
```

# Step 7: Hypotheses

**Null Hypothesis**

The average house value is the same.

H0​:μclose​=μfar​

**Alternative Hypothesis**

Houses close to a school or hospital have higher prices.

H1​:μclose​>μfar​


Since these are two independent groups, the correct test isIndependent Two-Sample T-Test (Welch's)

```python
# Step 8: performing the Test

t_stat, p_two = st.ttest_ind(close_prices, far_prices, equal_var=False)

# One-sided p-value
p_value = p_two / 2

print("T-statistic:", t_stat)
print("One-tailed p-value:", p_value)
```

```python
# Step 8: Decision
alpha = 0.05

if (p_value < alpha) and (t_stat > 0):
    print("Reject H₀")
else:
    print("Fail to Reject H₀")
```

### Interpretation

If the p-value is less than 0.05:

- We reject the null hypothesis.
- There is sufficient evidence that houses located within 0.50 units of either the school or hospital have a higher average value.

If the p-value is greater than 0.05:

- We fail to reject the null hypothesis.
- There is insufficient evidence to conclude that houses close to a school or hospital are more expensive.



Interpretation (α = 0.05)
If p < 0.05, houses close to school/hospital are more expensive.

If p ≥ 0.05, no evidence of higher prices.

Expected result:  
Areas near hospitals/schools tend to be more urban → higher prices.

📝 Conclusion
At the 5% significance level, we reject the null hypothesis.
Houses close to a school or hospital have significantly higher median values than houses farther away.
