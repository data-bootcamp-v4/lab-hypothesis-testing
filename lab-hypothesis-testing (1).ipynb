In this challenge, we will be working with pokemon data. The data can be found here:

```python
#libraries
import pandas as pd
import scipy.stats as st
import numpy as np

```

```python
df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv")
df
```

- We posit that Pokemons of type Dragon have, on average, more HP stats than Grass. Choose the propper test and, with 5% significance, comment your findings.

```python
# ---------------------------------------------------------------
# QUESTION 1: Do Dragon Pokemons have MORE HP than Grass Pokemons?
# ---------------------------------------------------------------
# We compare the means of 2 DIFFERENT groups (Dragon vs Grass)
# -> the groups are independent -> Two Sample T-Test (st.ttest_ind)
#
# H0: mean HP Dragon <= mean HP Grass
# H1: mean HP Dragon >  mean HP Grass   (one-sided test -> alternative='greater')
# Significance level alpha = 0.05

# 1. Select the HP of each group
dragon_hp = df[df["Type 1"] == "Dragon"]["HP"]
grass_hp = df[df["Type 1"] == "Grass"]["HP"]

# 2. Look at the means first (just to have an idea)
print("Mean HP Dragon:", dragon_hp.mean())
print("Mean HP Grass :", grass_hp.mean())

# 3. Run the test
# equal_var=False -> we don't assume both groups have the same variance 
t_stat, p_value = st.ttest_ind(dragon_hp, grass_hp, equal_var=False, alternative="greater")
print("t-statistic:", t_stat)
print("p-value    :", p_value)

# 4. Decision: compare p-value with alpha
alpha = 0.05
if p_value < alpha:
    print("p-value < 0.05 -> we REJECT H0: Dragon Pokemons have on average more HP than Grass Pokemons.")
else:
    print("p-value >= 0.05 -> we FAIL to reject H0: no evidence that Dragon has more HP than Grass.")
```

**Comment:** we used a Two Sample T-Test because Dragon and Grass are two independent groups. The test is one-sided because the hypothesis says "more HP". If the p-value is lower than 0.05 we reject H0 and conclude that Dragon Pokemons have, on average, more HP than Grass Pokemons.

- We posit that Legendary Pokemons have different stats (HP, Attack, Defense, Sp.Atk, Sp.Def, Speed) when comparing with Non-Legendary. Choose the propper test and, with 5% significance, comment your findings.


```python
# ---------------------------------------------------------------
# QUESTION 2: Do Legendary Pokemons have DIFFERENT stats than Non-Legendary?
# ---------------------------------------------------------------
# Again 2 independent groups (Legendary vs Non-Legendary) -> Two Sample T-Test
# This time the hypothesis says "different" (not "more") -> two-sided test (default)
#
# For EACH stat:
# H0: mean stat Legendary == mean stat Non-Legendary
# H1: mean stat Legendary != mean stat Non-Legendary
# alpha = 0.05

# 1. Split the data into the 2 groups
legendary = df[df["Legendary"] == True]
non_legendary = df[df["Legendary"] == False]

# 2. List of stats we want to test
stats = ["HP", "Attack", "Defense", "Sp. Atk", "Sp. Def", "Speed"]

alpha = 0.05

# 3. Loop: one t-test for each stat
for stat in stats:
    t_stat, p_value = st.ttest_ind(legendary[stat], non_legendary[stat], equal_var=False)

    # decision for this stat
    if p_value < alpha:
        decision = "Reject H0 -> different"
    else:
        decision = "Fail to reject H0 -> no difference"

    print(stat, "| mean Legendary:", round(legendary[stat].mean(), 1),
          "| mean Non-Legendary:", round(non_legendary[stat].mean(), 1),
          "| p-value:", p_value, "|", decision)
```

**Comment:** for each stat we ran a two-sided Two Sample T-Test. When the p-value is lower than 0.05 for a stat, we reject H0: Legendary and Non-Legendary Pokemons have different means for that stat (and looking at the means, Legendary ones are higher).

In this challenge, we will be working with california-housing data. The data can be found here:

```python
df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/california_housing.csv")
df.head()
```

**We posit that houses close to either a school or a hospital are more expensive.**

- School coordinates (-118, 34)
- Hospital coordinates (-122, 37)

We consider a house (neighborhood) to be close to a school or hospital if the distance is lower than 0.50.

Hint:
- Write a function to calculate euclidean distance from each house (neighborhood) to the school and to the hospital.
- Divide your dataset into houses close and far from either a hospital or school.
- Choose the propper test and, with 5% significance, comment your findings.
 

```python
# ---------------------------------------------------------------
# CHALLENGE 2: Are houses close to a school or a hospital more expensive?
# ---------------------------------------------------------------

# 1. Function to calculate the euclidean distance between 2 points
#    distance = square root of ((x1 - x2)^2 + (y1 - y2)^2)
def euclidean_distance(x1, y1, x2, y2):
    return np.sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)

# Coordinates given in the exercise (longitude, latitude)
school = (-118, 34)
hospital = (-122, 37)

# 2. Distance from each house to the school and to the hospital (new columns)
df["dist_school"] = euclidean_distance(df["longitude"], df["latitude"], school[0], school[1])
df["dist_hospital"] = euclidean_distance(df["longitude"], df["latitude"], hospital[0], hospital[1])

# 3. A house is "close" if it is at less than 0.50 from the school OR from the hospital
df["close"] = (df["dist_school"] < 0.5) | (df["dist_hospital"] < 0.5)

# 4. Divide the dataset into 2 groups
close_houses = df[df["close"] == True]["median_house_value"]
far_houses = df[df["close"] == False]["median_house_value"]

print("Number of close houses:", len(close_houses))
print("Number of far houses  :", len(far_houses))
print("Mean value close:", close_houses.mean())
print("Mean value far  :", far_houses.mean())
```

```python
# Close houses and far houses are 2 independent groups -> Two Sample T-Test
# The hypothesis says "more expensive" -> one-sided test (alternative='greater')
#
# H0: mean price close <= mean price far
# H1: mean price close >  mean price far
# alpha = 0.05

t_stat, p_value = st.ttest_ind(close_houses, far_houses, equal_var=False, alternative="greater")
print("t-statistic:", t_stat)
print("p-value    :", p_value)

alpha = 0.05
if p_value < alpha:
    print("p-value < 0.05 -> we REJECT H0: houses close to a school or hospital are more expensive.")
else:
    print("p-value >= 0.05 -> we FAIL to reject H0: no evidence that close houses are more expensive.")
```

**Comment:** we split the neighborhoods into "close" (distance < 0.50 from the school or the hospital) and "far", then used a one-sided Two Sample T-Test. If the p-value is lower than 0.05 we reject H0 and conclude that houses close to a school or a hospital are, on average, more expensive.
