**Challenge 1**

In this challenge, we will be working with pokemon data. The data can be found here:

- https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv

```python
#libraries
import pandas as pd
import scipy.stats as st
import numpy as np
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv")
df
```

- We posit that Pokemons of type Dragon have, on average, more HP stats than Grass. Choose the propper test and, with 5% significance, comment your findings.

```python
# Challenge 1.1
# Question: Do Dragon-type Pokémon have higher HP than Grass-type Pokémon?

dragon_hp = df[df["Type 1"] == "Dragon"]["HP"]
grass_hp = df[df["Type 1"] == "Grass"]["HP"]

print("Dragon sample size:", len(dragon_hp))
print("Grass sample size:", len(grass_hp))
print("Dragon average HP:", round(dragon_hp.mean(), 2))
print("Grass average HP:", round(grass_hp.mean(), 2))

# Hypotheses:
# H0: mean HP of Dragon <= mean HP of Grass
# H1: mean HP of Dragon > mean HP of Grass
#
# We use a two-sample independent t-test because:
# - Dragon and Grass are two different groups
# - We compare their average HP
# - We do not assume equal variance between the groups

t_stat, p_value = st.ttest_ind(
    dragon_hp,
    grass_hp,
    equal_var=False,
    alternative="greater"
)

print("\nt-statistic:", round(t_stat, 4))
print("p-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
    print("Conclusion: There is enough statistical evidence to say that Dragon-type Pokémon have higher average HP than Grass-type Pokémon.")
else:
    print("Fail to reject H0")
    print("Conclusion: There is not enough statistical evidence to say that Dragon-type Pokémon have higher average HP than Grass-type Pokémon.")
```

**Interpretation**

For this question, we use an **independent two-sample t-test** because we are comparing the average HP of two separate groups: Dragon-type Pokémon and Grass-type Pokémon.

Since the business/statistical question says *"Dragon Pokémon have more HP than Grass Pokémon"*, this is a **one-tailed test** with `alternative="greater"`.

- If `p-value < 0.05`, we reject the null hypothesis.
- If `p-value >= 0.05`, we fail to reject the null hypothesis.

In simple words: we are checking whether the difference in average HP is big enough to believe it is not just random variation.

- We posit that Legendary Pokemons have different stats (HP, Attack, Defense, Sp.Atk, Sp.Def, Speed) when comparing with Non-Legendary. Choose the propper test and, with 5% significance, comment your findings.


```python
df.columns
```

```python
# Challenge 1.2
# Question: Do Legendary Pokémon have different stats compared with Non-Legendary Pokémon?

stats_cols = [
    "HP",
    "Attack",
    "Defense",
    "Sp. Atk",
    "Sp. Def",
    "Speed"
]

alpha = 0.05
results = []

for col in stats_cols:

    legendary = df[df["Legendary"] == True][col]
    non_legendary = df[df["Legendary"] == False][col]

    # Hypotheses:
    # H0: mean stat of Legendary = mean stat of Non-Legendary
    # H1: mean stat of Legendary != mean stat of Non-Legendary
    #
    # We use a two-sample independent t-test because:
    # - Legendary and Non-Legendary are two independent groups
    # - We compare the mean of each stat
    # - We are testing if they are different, not specifically greater or lower

    t_stat, p_value = st.ttest_ind(
        legendary,
        non_legendary,
        equal_var=False,
        alternative="two-sided"
    )

    decision = "Reject H0" if p_value < alpha else "Fail to reject H0"

    results.append({
        "Stat": col,
        "Legendary Mean": round(legendary.mean(), 2),
        "Non-Legendary Mean": round(non_legendary.mean(), 2),
        "t-statistic": round(t_stat, 4),
        "p-value": p_value,
        "Decision": decision
    })

results_df = pd.DataFrame(results)
results_df
```

**Interpretation**

For this question, we test each stat separately: HP, Attack, Defense, Sp. Atk, Sp. Def and Speed.

We use an **independent two-sample t-test** because Legendary and Non-Legendary Pokémon are two different groups.

This time, the question says *"have different stats"*, not *"higher stats"*. That means we use a **two-tailed test**:

- `H0`: Legendary and Non-Legendary Pokémon have the same average stat.
- `H1`: Legendary and Non-Legendary Pokémon have different average stats.

For every stat where `p-value < 0.05`, we reject the null hypothesis and conclude that there is a statistically significant difference between Legendary and Non-Legendary Pokémon for that stat.

**We posit that houses close to either a school or a hospital are more expensive.**

- School coordinates (-118, 34)
- Hospital coordinates (-122, 37)

We consider a house (neighborhood) to be close to a school or hospital if the distance is lower than 0.50.

Hint:
- Write a function to calculate euclidean distance from each house (neighborhood) to the school and to the hospital.
- Divide your dataset into houses close and far from either a hospital or school.
- Choose the propper test and, with 5% significance, comment your findings.
 

```python
# Challenge 2
# Load California housing data

housing = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/california_housing.csv")
housing.head()
```

```python
# Coordinates
school = (-118, 34)
hospital = (-122, 37)

# Function to calculate Euclidean distance
def euclidean_distance(lon, lat, point):
    point_lon, point_lat = point
    return np.sqrt((lon - point_lon)**2 + (lat - point_lat)**2)

# Calculate distance from each house/neighborhood to school and hospital
housing["distance_to_school"] = euclidean_distance(
    housing["longitude"],
    housing["latitude"],
    school
)

housing["distance_to_hospital"] = euclidean_distance(
    housing["longitude"],
    housing["latitude"],
    hospital
)

# A house is close if it is within 0.50 distance units from either the school or the hospital
housing["close_to_school_or_hospital"] = (
    (housing["distance_to_school"] < 0.50) |
    (housing["distance_to_hospital"] < 0.50)
)

housing[["longitude", "latitude", "median_house_value", "distance_to_school", "distance_to_hospital", "close_to_school_or_hospital"]].head()
```

```python
# Split the data into two independent groups

close_houses = housing[housing["close_to_school_or_hospital"] == True]["median_house_value"]
far_houses = housing[housing["close_to_school_or_hospital"] == False]["median_house_value"]

print("Close houses sample size:", len(close_houses))
print("Far houses sample size:", len(far_houses))
print("Average value of close houses:", round(close_houses.mean(), 2))
print("Average value of far houses:", round(far_houses.mean(), 2))
```

```python
# Hypotheses:
# H0: mean price of close houses <= mean price of far houses
# H1: mean price of close houses > mean price of far houses
#
# We use a two-sample independent t-test because:
# - close houses and far houses are two different groups
# - we compare their average median house value
# - the question says "close houses are more expensive", so this is a one-tailed test

alpha = 0.05

t_stat, p_value = st.ttest_ind(
    close_houses,
    far_houses,
    equal_var=False,
    alternative="greater"
)

print("t-statistic:", round(t_stat, 4))
print("p-value:", p_value)

if p_value < alpha:
    print("Reject H0")
    print("Conclusion: There is enough statistical evidence to say that houses close to a school or hospital are more expensive.")
else:
    print("Fail to reject H0")
    print("Conclusion: There is not enough statistical evidence to say that houses close to a school or hospital are more expensive.")
```

**Interpretation**

For this challenge, we create two groups:

1. Houses close to either the school or the hospital.
2. Houses far from both places.

Because these are two independent groups and we compare their average house value, the proper test is an **independent two-sample t-test**.

The hypothesis says that close houses are **more expensive**, so we use a **one-tailed test** with `alternative="greater"`.

- `H0`: houses close to a school or hospital are not more expensive.
- `H1`: houses close to a school or hospital are more expensive.

If the `p-value` is below `0.05`, we reject `H0`. That would mean the observed price difference is unlikely to be just random noise.
