# Hypothesis Testing Lab — Challenge 1 (Pokemon)

Two questions, each compared at the **5% significance level (α = 0.05)**:

1. Do **Dragon** Pokemon have, on average, *more* HP than **Grass** Pokemon? (one-sided)
2. Do **Legendary** Pokemon have *different* stats than **Non-Legendary** Pokemon, across HP, Attack, Defense, Sp. Atk, Sp. Def, Speed? (two-sided, per stat)

```python
import pandas as pd
from scipy.stats import ttest_ind

df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv")
df.head()
```

## Task 1 — Dragon HP vs Grass HP

We compare the mean HP of two **independent** groups, so the proper test is an **independent two-sample t-test**. The claim is *directional* (Dragon > Grass), so we use a **one-tailed** test.

- **H0:** mean HP(Dragon) ≤ mean HP(Grass)
- **H1:** mean HP(Dragon) > mean HP(Grass)

We use Welch's t-test (`equal_var=False`) since the two groups have different sizes and we don't assume equal variances. `scipy`'s `ttest_ind` returns a two-sided p-value, so for the one-sided test we halve it when the sample difference is in the hypothesized direction.

```python
dragon_hp = df[df["Type 1"] == "Dragon"]["HP"]
grass_hp  = df[df["Type 1"] == "Grass"]["HP"]

print(f"Dragon: n={dragon_hp.size}, mean HP={dragon_hp.mean():.2f}")
print(f"Grass : n={grass_hp.size}, mean HP={grass_hp.mean():.2f}")

t_stat, p_two_sided = ttest_ind(dragon_hp, grass_hp, equal_var=False)
p_one_sided = p_two_sided / 2 if t_stat > 0 else 1 - p_two_sided / 2

print(f"\nt-statistic     = {t_stat:.4f}")
print(f"p-value (1-tail) = {p_one_sided:.4f}")
print("Reject H0" if p_one_sided < 0.05 else "Fail to reject H0")
```

### Findings — Task 1

The one-tailed p-value is well below 0.05, so we **reject H0**. There is statistically significant evidence at the 5% level that Dragon-type Pokemon have, on average, **more HP** than Grass-type Pokemon (sample means confirm Dragon HP is the higher of the two).

## Task 2 — Legendary vs Non-Legendary stats

For each stat we compare two **independent** groups and ask whether they *differ* (no direction stated), so the proper test is an **independent two-sample t-test, two-tailed**, run once per stat.

- **H0:** mean(Legendary) = mean(Non-Legendary)
- **H1:** mean(Legendary) ≠ mean(Non-Legendary)

```python
stats = ["HP", "Attack", "Defense", "Sp. Atk", "Sp. Def", "Speed"]

legendary = df[df["Legendary"] == True]
non_legendary = df[df["Legendary"] == False]

results = []
for s in stats:
    t_stat, p_value = ttest_ind(legendary[s], non_legendary[s], equal_var=False)
    results.append({
        "stat": s,
        "mean_legendary": legendary[s].mean(),
        "mean_non_legendary": non_legendary[s].mean(),
        "t_stat": t_stat,
        "p_value": p_value,
        "significant (p<0.05)": p_value < 0.05,
    })

pd.DataFrame(results).set_index("stat")
```

### Findings — Task 2

Every stat returns a p-value far below 0.05, so we **reject H0 for all six stats**. At the 5% significance level, Legendary Pokemon have statistically significant **different** (and consistently higher) HP, Attack, Defense, Sp. Atk, Sp. Def, and Speed than Non-Legendary Pokemon.

# Challenge 2 — California Housing

Claim: houses close to **either** a school or a hospital are *more expensive*.

- School at (-118, 34), Hospital at (-122, 37)
- A house is "close" if its **euclidean distance** to the school **or** hospital is **< 0.50**
- Compare `median_house_value` of close vs far houses at α = 0.05

```python
import numpy as np

df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/california_housing.csv")

def euclidean(lon, lat, point):
    return np.sqrt((lon - point[0]) ** 2 + (lat - point[1]) ** 2)

df["dist_school"]   = euclidean(df["longitude"], df["latitude"], (-118, 34))
df["dist_hospital"] = euclidean(df["longitude"], df["latitude"], (-122, 37))
df["close"] = (df["dist_school"] < 0.50) | (df["dist_hospital"] < 0.50)

df[["longitude", "latitude", "dist_school", "dist_hospital", "close"]].head()
```

We split into two **independent** groups (close vs far) and the claim is *directional* (close > far), so we use an **independent two-sample t-test, one-tailed**.

- **H0:** mean value(close) ≤ mean value(far)
- **H1:** mean value(close) > mean value(far)

```python
close = df[df["close"]]["median_house_value"]
far   = df[~df["close"]]["median_house_value"]

print(f"Close: n={close.size}, mean value={close.mean():.0f}")
print(f"Far  : n={far.size}, mean value={far.mean():.0f}")

t_stat, p_two_sided = ttest_ind(close, far, equal_var=False)
p_one_sided = p_two_sided / 2 if t_stat > 0 else 1 - p_two_sided / 2

print(f"\nt-statistic     = {t_stat:.4f}")
print(f"p-value (1-tail) = {p_one_sided:.4e}")
print("Reject H0" if p_one_sided < 0.05 else "Fail to reject H0")
```

### Findings — Challenge 2

The one-tailed p-value is essentially 0 (≪ 0.05), so we **reject H0**. There is statistically significant evidence at the 5% level that houses close to a school or hospital are **more expensive**: the close group averages ~$246,952 (n=6829) vs ~$180,678 (n=10171) for the far group.
