- We posit that Pokemons of type Dragon have, on average, more HP stats than Grass. Choose the propper test and, with 5% significance, comment your findings.

```python
url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv"
pokemon_data = pd.read_csv(url)
```

```python
from scipy.stats import ttest_ind

dragon_hp = pokemon_data[pokemon_data['Type 1'] == 'Dragon']['HP']
grass_hp = pokemon_data[pokemon_data['Type 1'] == 'Grass']['HP']

t_stat, p_value = ttest_ind(dragon_hp, grass_hp, alternative='greater')

alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: Dragon-type Pokémon have higher average HP than Grass-type.")
else:
    print("Fail to reject the null hypothesis: No significant difference in average HP between Dragon and Grass Pokémon.")
```

- We posit that Legendary Pokemons have different stats (HP, Attack, Defense, Sp.Atk, Sp.Def, Speed) when comparing with Non-Legendary. Choose the propper test and, with 5% significance, comment your findings.


```python
legendary_stats = pokemon_data[pokemon_data['Legendary'] == True]
non_legendary_stats = pokemon_data[pokemon_data['Legendary'] == False]

stats = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']

alpha = 0.05
results = {}

for stat in stats:
    stat_legendary = legendary_stats[stat]
    stat_non_legendary = non_legendary_stats[stat]
    t_stat, p_value = ttest_ind(stat_legendary, stat_non_legendary, equal_var=False)

    results[stat] = {
        't-statistic': t_stat,
        'p-value': p_value,
        'significant': p_value < alpha
    }

for stat, result in results.items():
    print(f"{stat}:")
    print(f"  t-statistic: {result['t-statistic']:.3f}")
    print(f"  p-value: {result['p-value']:.3f}")
    print(f"  Significant difference: {result['significant']}")
    print("---")
```

**We posit that houses close to either a school or a hospital are more expensive.**

- School coordinates (-118, 34)
- Hospital coordinates (-122, 37)

We consider a house (neighborhood) to be close to a school or hospital if the distance is lower than 0.50.

Hint:
- Write a function to calculate euclidean distance from each house (neighborhood) to the school and to the hospital.
- Divide your dataset into houses close and far from either a hospital or school.
- Choose the propper test and, with 5% significance, comment your findings.


```python
url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/california_housing.csv"
housing_data = pd.read_csv(url)
housing_data.head()
```

Euclidean distance will have its own formula vs housing_data['distance_to_school'] = np.sqrt((housing_data['longitude'] - school_coords[0])**2 + (housing_data['latitude'] - school_coords[1])**2)

```python
def euclidean_distance(point1, point2):
    return np.sqrt((point1[0] - point2[0])**2 + (point1[1] - point2[1])**2)
```

```python
# Coordinates for school and hospital
school_coords = (-118, 34)
hospital_coords = (-122, 37)

# Distances
housing_data['distance_to_school'] = housing_data.apply(lambda row: euclidean_distance((row['longitude'], row['latitude']), school_coords), axis=1)
housing_data['distance_to_hospital'] = housing_data.apply(lambda row: euclidean_distance((row['longitude'], row['latitude']), hospital_coords), axis=1)
```

```python
housing_data['close'] = (housing_data['distance_to_school'] < 0.50) | (housing_data['distance_to_hospital'] < 0.50)
```

```python
close_group = housing_data[housing_data['close']]['median_house_value']
far_group = housing_data[~housing_data['close']]['median_house_value']
```

```python
t_stat, p_value = ttest_ind(close_group, far_group, equal_var=False)
```

```python
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: Houses close to a school or hospital are more expensive on average.")
else:
    print("Fail to reject the null hypothesis: No significant difference in house prices based on proximity to school or hospital.")
```
