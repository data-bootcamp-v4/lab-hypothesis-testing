```python
import pandas as pd
import scipy.stats as st
import numpy as np
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv")

df
```

```python
df["Type 1"].unique()
```

```python
df["Type 2"].unique()
```

```python
# H0 = Average HP Dragon <= Average HP Grass
# H1 = Average HP Dragon > Average HP Grass
```

```python
alpha = 0.05
```

```python
dragon_hp = df[(df['Type 1'] == 'Dragon') | (df['Type 2'] == 'Dragon')]['HP']
grass_hp = df[(df['Type 1'] == 'Grass') | (df['Type 2'] == 'Grass')]['HP']
```

```python
_, p_value = st.ttest_ind(dragon_hp, grass_hp, alternative='greater'
                          , equal_var=False)

print(p_value)
```

```python
if p_value < alpha:
    print("Conclusion: Reject H0. There is statistically significant evidence that Dragon-type Pokémon have, on average, more HP stats than Grass-type.")
else:
    print("Conclusion: Fail to reject H0. There is insufficient evidence to conclude that Dragon-type Pokémon have higher average HP than Grass-type.")
```

```python
# Ho = stats Legendary (HP, Attack, Defense, Sp.Atk, Sp.Def, Speed) = stats Non-Legendary (HP, Attack, Defense, Sp.Atk, Sp.Def, Speed)
# H1 = stats Legendary (HP, Attack, Defense, Sp.Atk, Sp.Def, Speed) ≠ stats Non-Legendary (HP, Attack, Defense, Sp.Atk, Sp.Def, Speed)
```

```python
stats_to_test = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']
```

```python
legendary_df = df[df['Legendary'] == True]
non_legendary_df = df[df['Legendary'] == False]
```

```python
alpha = 0.05
```

```python
for stat in stats_to_test:
    leg_stats = legendary_df[stat]
    non_leg_stats = non_legendary_df[stat]

    _, p_value = st.ttest_ind(leg_stats, non_leg_stats, alternative='two-sided')

    print(p_value)


    if p_value < alpha:
        print(f"Conclusion: Reject H0. There is a statistically significant difference in {stat} between Legendary and Non-Legendary Pokémon.\n")
    else:
        print(f"Conclusion: Fail to reject H0. No significant difference found in {stat}.\n")

```

```python
df_2 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/california_housing.csv")

df_2.head()
```

```python
def calculate_distances(df_2):
  # Define coordinates for school and hospital
  school_coords = (-118, 34)
  hospital_coords = (-122, 37)

  # Calculate Euclidean distance to the school
  df_2["dist_to_school"] = np.sqrt(
      (df_2["longitude"] - school_coords[0]) ** 2
      + (df_2["latitude"] - school_coords[1]) ** 2
  )

  # Calculate Euclidean distance to the hospital
  df_2["dist_to_hospital"] = np.sqrt(
      (df_2["longitude"] - hospital_coords[0]) ** 2
      + (df_2["latitude"] - hospital_coords[1]) ** 2
  )

  return df_2
```

```python
df_2 = calculate_distances(df_2)

# Define a threshold for being "close" (distance < 0.50)
threshold = 0.50

# Divide the dataset into houses close to either a school or a hospital, and those far away
houses_close = df_2[
    (df_2["dist_to_school"] < threshold) | (df_2["dist_to_hospital"] < threshold)
]
houses_far = df_2[
    (df_2["dist_to_school"] >= threshold)
    & (df_2["dist_to_hospital"] >= threshold)
]
```

```python
# H0 : mu_close <= mu_far (Close houses are not more expensive)
# H1:  mu_close > mu_far (Close houses are more expensive)
```

```python
# Extract the median house value for each group
close_values = houses_close['median_house_value'].dropna()
far_values = houses_far['median_house_value'].dropna()
```

```python
# Perform One-tailed Independent Two-Sample T-Test (alternative='greater')
_, p_value = st.ttest_ind(close_values, far_values, alternative='greater')

print(p_value)
```

```python
alpha = 0.05
if p_value < alpha:
  print(
      "Conclusion: Reject H0. There is statistically significant evidence that"
      " houses close to a school or hospital are more expensive."
  )
else:
  print(
      "Conclusion: Fail to reject H0. There is insufficient evidence to"
      " conclude that houses close to a school or hospital are more expensive."
  )
```
