```python
import pandas as pd
import scipy.stats as st
import numpy as np
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv")
df
```

```python
### h0: mean hp of dragon == mean hp of grass 
### h1: mean hp of dragon != mean hp of grass 
```

```python
from scipy.stats import ttest_ind
```

```python
hp_dragon = df[df['Type 1'] == 'Dragon']['HP']
hp_grass = df[df['Type 1'] == 'Grass']['HP']

t_stat, p_val = ttest_ind(hp_dragon, hp_grass, equal_var=False, alternative='greater')

print("t-statistic:", t_stat)
print("p-value:", p_val)
```

```python
### reject the null hypothesis, there is statistical evidence to support that dragon type has higher HP 
```

```python
### h0: mean stats of legendary == stats of non-legendary 
### h1: mean stats of legendary != stats of non-legendary
```

```python
stats_columns = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']

legendary = df[df['Legendary'] == True]
non_legendary = df[df['Legendary'] == False]

results = {}
for col in stats_columns:
    t_stat, p_val = ttest_ind(legendary[col], non_legendary[col], equal_var=False)
    results[col] = {'t-statistic': t_stat, 'p-value': p_val}

pd.DataFrame(results)
```

```python
### reject the null hypothesis, there is statistical evidence to support that legendary pokemon
### have higher stats in all categories than non-legendary pokemon
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/california_housing.csv")
df.head()
```

```python
### h0: the mean median_house_value close to school/hospital is == the mean median_house_value far to school/hospital 
### h1: he mean median_house_value close to school/hospital is != the mean median_house_value far to school/hospital
```

```python
import numpy as np

### coords
school = (-118, 34)
hospital = (-122, 37)

### function for euclidean distance
def euclidean_distance(lon, lat, point):
    return np.sqrt((lon - point[0])**2 + (lat - point[1])**2)

### distances to school and hospital
df['dist_school'] = euclidean_distance(df['longitude'], df['latitude'], school)
df['dist_hospital'] = euclidean_distance(df['longitude'], df['latitude'], hospital)

### finding out if house is "close" (<0.50)
df['close_to_facility'] = ((df['dist_school'] < 0.50) | (df['dist_hospital'] < 0.50))
```

```python
### split df to close and far
close_houses = df[df['close_to_facility'] == True]['median_house_value']
far_houses = df[df['close_to_facility'] == False]['median_house_value']
```

```python
t_stat, p_val = ttest_ind(close_houses, far_houses, equal_var=False, alternative='greater')

print("t-statistic:", t_stat)
print("p-value:", p_val)
```

```python
### reject the null hypothesis, there is really strong statistical evidence to support that houses close to a school or hospital are 
### much more expensive than houses farther away
```
