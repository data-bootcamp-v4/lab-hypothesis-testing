**Challenge 1**

In this challenge, we will be working with pokemon data. The data can be found here:

- https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv

```python
#libraries
import pandas as pd
import scipy.stats as st
import numpy as np

```

```python
df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/pokemon.csv")
df
```

- We posit that Pokemons of type Dragon have, on average, more HP stats than Grass. Choose the propper test and, with 5% significance, comment your findings.

```python
#code here
s = df[df["Type 1"] == "Dragon"]["HP"]
t = df[df["Type 1"] == "Grass"]["HP"]
st.ttest_ind(s, t, equal_var=False)
# Since the p-value (0.0016) is less than the significance level (0.05), we reject the null hypothesis.
#The t-statistic is positive (3.33), which supports the direction of the alternative hypothesis (Dragon HP > Grass HP).
```

- We posit that Legendary Pokemons have different stats (HP, Attack, Defense, Sp.Atk, Sp.Def, Speed) when comparing with Non-Legendary. Choose the propper test and, with 5% significance, comment your findings.


```python
#code here
s = df[df["Legendary"] == True][["HP", "Attack", "Defense", "Sp. Atk", "Sp. Def", "Speed"]]
t = df[df["Legendary"] == False][["HP", "Attack", "Defense", "Sp. Atk", "Sp. Def", "Speed"]]
st.ttest_ind(s, t, equal_var=False)
#reject $H_0$$H_0$. There is a significant difference in average Speed between Legendary and Non-Legendary Pokémon.
```

In this challenge, we will be working with california-housing data. The data can be found here:
- https://raw.githubusercontent.com/data-bootcamp-v4/data/main/california_housing.csv

```python
df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/california_housing.csv")
df.head()
```

**We posit that houses close to either a school or a hospital are more expensive.**

- School coordinates (-118, 34)
- Hospital coordinates (-122, 37)

We consider a house (neighborhood) to be close to a school or hospital if the distance is lower than 0.50.

Hint:
- Write a function to calculate euclidean distance from each house (neighborhood) to the school and to the hospital.
- Divide your dataset into houses close and far from either a hospital or school.
- Choose the propper test and, with 5% significance, comment your findings.


```python
import pandas as pd
import numpy as np
from scipy.stats import ttest_ind, mannwhitneyu

school = (-118, 34)
hospital = (-122, 37)

def euclidean_dist(coord1, coord2):
    return np.sqrt((coord1[0] - coord2[0])**2 + (coord1[1] - coord2[1])**2)

df['distance_to_school'] = df.apply(lambda row: euclidean_dist((row['longitude'], row['latitude']), school), axis=1)
df['distance_to_hospital'] = df.apply(lambda row: euclidean_dist((row['longitude'], row['latitude']), hospital), axis=1)

df['is_close'] = (df['distance_to_school'] < 0.5) | (df['distance_to_hospital'] < 0.5)

# Split data
close_prices = df[df['is_close']]['median_house_value']
far_prices = df[~df['is_close']]['median_house_value']

# Check for normality first (optional)
# If not normal, use Mann-Whitney
stat, p_value = ttest_ind(close_prices, far_prices, alternative='greater', equal_var=False)

if p_value < 0.05:
    print("✅ Reject H0: Houses close to school/hospital are significantly more expensive.")
else:
    print("❌ Do not reject H0: No significant price difference found.")
```
